```python
import sys
from pathlib import Path
import ase
from pymatgen.core import Structure

print("Python:", sys.version.split()[0])
print("Python location:", sys.executable)
print("ASE:", ase.__version__)
print("CIF found:", Path("data/catio3_bulk.cif").exists())
```

```python
from pathlib import Path
from pymatgen.core import Structure

cif_path = Path("data") / "catio3_bulk.cif"

print("Notebook is running from:", Path.cwd())
print("Looking for:", cif_path.resolve())
print("CIF found:", cif_path.is_file())
```

```python
structure = Structure.from_file(str(cif_path))
print("Reduced formula:", structure.composition.reduced_formula)
print("Number of sites:", len(structure))
```

```python
print("Lattice vectors (Å):")
print(structure.lattice.matrix)

print("Cell lengths (Å):", structure.lattice.abc)
print("Cell angles (degrees):", structure.lattice.angles)
print("Cell volume (Å³):", structure.volume)
```

```python
for site in structure:
    print(site.species_string, site.frac_coords)
```

```python
symbol, number = structure.get_space_group_info()
print("Space group:", symbol, "(", number, ")")
```

```python
from ase.io import read

ase_atoms = read(str(cif_path))

print("Formula:", ase_atoms.get_chemical_formula())
print("Number of atoms:", len(ase_atoms))
print("Cell lengths (Å):", ase_atoms.cell.lengths())
print("Cell angles (degrees):", ase_atoms.cell.angles())
print("Periodicity:", ase_atoms.pbc)
```

```python
from ase.visualize import view

view(ase_atoms, viewer="x3d")
```

```python
from pathlib import Path

supercell = structure * (2, 2, 2)

print("Original sites:", len(structure))
print("Supercell sites:", len(supercell))
print("Supercell lengths (Å):", supercell.lattice.abc)

output_dir = Path("outputs")
output_dir.mkdir(parents=True, exist_ok=True)

supercell_cif = output_dir / "catio3_supercell.cif"
supercell.to(fmt="cif", filename=str(supercell_cif))

print("Saved CIF:", supercell_cif.resolve())
```

```python
from ase.io import read

supercell_atoms = read(str(supercell_cif))

print("CIF exists:", supercell_cif.is_file())
print("Formula:", supercell_atoms.get_chemical_formula())
print("Number of atoms:", len(supercell_atoms))
print("Cell lengths (Å):", supercell_atoms.cell.lengths())
```

```python
from ase.visualize import view

view(supercell_atoms, viewer="x3d")
```
